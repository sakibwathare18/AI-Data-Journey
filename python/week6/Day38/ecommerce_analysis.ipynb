```python
import pandas as pd
import os
```

```python
os.makedirs("output", exist_ok=True)
```

```python
df = pd.read_csv("ecommerce_data.csv")
```

```python
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
df.head()
```

Data Cleaning

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
print("\nMissing Values:")
print(df.isna().sum())
```

```python
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

Calculate Revenue

```python
# GROSS SALES
df["Gross_Sales"] = df["Quantity"] * df["Unit_Price"]
```

```python
# DISCOUNT AMOUNT
df["Discount_Amount"] = df["Gross_Sales"] * df["Discount"]
```

```python
# FINAL SALES
df["Net_Sales"] = df["Gross_Sales"] - df["Discount_Amount"]
```

```python
df[
    [
        "Order_ID",
        "Product",
        "Quantity",
        "Gross_Sales",
        "Discount_Amount",
        "Net_Sales"
    ]
]
```

Create Time Features

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.month_name()
df["Quarter"] = df["Order_Date"].dt.quarter
df["Day_Name"] = df["Order_Date"].dt.day_name()
```

Basic Business KPIs

```python
# TOTAL SALES
total_sales = df["Net_Sales"].sum()
```

```python
# TOTAL ORDERS
total_order = df["Order_ID"].sum()
```

```python
# TOTAL QUANTITY
total_quantity = df["Quantity"].sum()
```

```python
# AVERAGE ORDER VALUE
average_order_value = df["Net_Sales"].mean()
```

```python
# AVERAGE RATING
average_rating = df["Rating"].mean()
```

```python
print("\n========== BUSINESS KPIs ==========")
print("Total Sales:", total_sales)
print("Total Orders:", total_order)
print("Total Quantity:", total_quantity)
print("Average Order Value:", average_order_value)
print("Average Rating:", average_rating)
```

City Analysis

```python
city_report = (
    df.groupby("City")
    .agg(
        Total_Sales = ("Net_Sales", "sum"),
        Total_Orders = ("Order_ID", "nunique"),
        Total_Quantity = ("Quantity", "sum"),
        Average_Rating = ("Rating", "mean")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)
```

```python
print("\n========== CITY REPORT ==========")
print(city_report)
```

```python
city_report.to_csv(
    "output/city_sales_report.csv",
    index=False
)
```

Product Analysis

```python
product_report = (
    df.groupby("Product")
    .agg(
        Total_Sales = ("Net_Sales", "sum"),
        Total_Orders = ("Order_ID", "nunique"),
        Total_Quantity = ("Quantity", "sum"),
        Average_Rating = ("Rating", "mean")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)
```

```python
print("\n========== PRODUCT REPORT ==========")
print(product_report)
```

```python
product_report.to_csv(
    "output/product_sales_report.csv",
    index=False
)
```

Category Analysis

```python
category_report = (
    df.groupby("Category")
    .agg(
        Total_Sales = ("Net_Sales", "sum"),
        Total_Orders = ("Order_ID", "nunique"),
        Total_Quantity = ("Quantity", "sum"),
        Average_Rating = ("Rating", "mean")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)
```

```python
print("\n========== CATEGORY REPORT ==========")
print(category_report)
```

```python
category_report.to_csv(
    "output/category_sales_report.csv",
    index=False
)
```

Monthly Sales

```python
monthly_report = (
    df.groupby(["Year", "Month", "Month_Name"])
      .agg(
          Total_Sales=("Net_Sales", "sum"),
          Total_Orders=("Order_ID", "nunique"),
          Total_Quantity=("Quantity", "sum")
      )
      .reset_index()
      .sort_values(["Year", "Month"])
)
```

```python
print("\n========== MONTHLY REPORT ==========")
print(monthly_report)
```

```python
monthly_report.to_csv(
    "output/monthly_sales_report.csv",
    index=False
)
```

Payment Analysis

```python
payment_report = (
    df.groupby("Payment")
      .agg(
          Total_Sales=("Net_Sales", "sum"),
          Total_Orders=("Order_ID", "nunique")
      )
      .reset_index()
      .sort_values("Total_Sales", ascending=False)
)
```

```python
print("\n========== PAYMENT REPORT ==========")
print(payment_report)
```

```python
payment_report.to_csv(
    "output/payment_report.csv",
    index=False
)
```

Find Top Performers

```python
top_product = (
    product_report
    .sort_values("Total_Sales", ascending=False)
    .iloc[0]
)

print("\nTop Product:")
print(top_product)
```

```python
top_city = city_report.iloc[0]

print("\nTop City:")
print(top_city)
```

```python
top_category = category_report.iloc[0]

print("\nTop Category:")
print(top_category)
```

Ranking

```python
product_report["Sales_Rank"] = (
    product_report["Total_Sales"]
    .rank(ascending=False, method="dense")
    .astype(int)
)
```

```python
product_report = product_report.sort_values("Sales_Rank")
```

Customer Analysis

```python
customer_report = (
    df.groupby("Customer")
      .agg(
          Total_Sales=("Net_Sales", "sum"),
          Total_Orders=("Order_ID", "nunique"),
          Total_Quantity=("Quantity", "sum"),
          Average_Order_Value=("Net_Sales", "mean")
      )
      .reset_index()
      .sort_values("Total_Sales", ascending=False)
)
```

```python
customer_report.to_csv(
    "output/customer_sales_report.csv",
    index=False
)
```

Customer Segmentation

```python
customer_report["Customer_Segment"] = pd.cut(
    customer_report["Total_Sales"],
    bins=[0, 20000, 50000, float("inf")],
    labels=["Low Value", "Medium Value", "High Value"]
)
```

```python
print("\nCustomer Segments:")
print(
    customer_report[
        ["Customer", "Total_Sales", "Customer_Segment"]
    ]
)
```

Category × City Analysis

```python
category_city = pd.pivot_table(
    df,
    index="Category",
    columns="City",
    values="Net_Sales",
    aggfunc="sum",
    fill_value=0,
    margins=True
)
```

```python
print("\n========== CATEGORY × CITY ==========")
print(category_city)
```

```python
category_city.to_csv(
    "output/category_city_matrix.csv"
)
```

Create an Executive Summary

```python
print("\n" + "=" * 50)
print("        E-COMMERCE EXECUTIVE SUMMARY")
print("=" * 50)

print(f"Total Sales       : ₹{total_sales:,.2f}")
print(f"Total Orders      : {total_order}")
print(f"Total Quantity    : {total_quantity}")
print(f"Average Order     : ₹{average_order_value:,.2f}")
print(f"Average Rating    : {average_rating:.2f}")

print(f"\nTop Product       : {top_product['Product']}")
print(f"Top City          : {top_city['City']}")
print(f"Top Category      : {top_category['Category']}")

print("=" * 50)
```
