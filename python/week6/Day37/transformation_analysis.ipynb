```python
import pandas as pd
```

```python
df = pd.read_csv("employees.csv")
df
```

```python
df["Department"] = df["Department"].map(
    str.upper
)
df["Department"]
```

```python
department_codes = {
    "Data": "DAT",
    "Engineering": "ENG",
    "Analytics": "ANA"
}
```

```python
df["Department"] = df["Department"].map(
    department_codes
)
```

```python
df
```

```python
df["City"] = df["City"].replace(
    {
        "Mumbai": "BOM",
        "Pune": "PNQ",
        "Delhi": "DEL"
    }
)
df["City"]
```

```python
df["Skill_Average"] = df[
    ["Python", "SQL", "ML"]
].mean(axis=1)

df["Skill_Average"] = (
    df["Skill_Average"].round(2)
)
```

```python
def performance(score):

    if score >= 90:
        return "Excellent"

    elif score >= 80:
        return "Good"
    
    elif score >= 70:
        return "Average"

    else:
        return "Needs Improvement"
```

```python
df["Performance"] = df[
    "Skill_Average"
].apply(performance)
```

```python
df[["Name", "Skill_Average", "Performance"]]
```

```python
df["Experience_Level"] = df[
    "Experience"
].apply(
    lambda x: "Senior" if x >= 5 else "Junior"
)
```

```python
df
```

```python
department_avg = df.groupby(
    "Department"
)["Salary"].transform("mean")
department_avg
```

```python
df["Department_Avg_Salary"] = (
    department_avg.round(2)
)
```

```python
df[
    [
        "Name",
        "Department",
        "Salary",
        "Department_Avg_Salary"
    ]
]
```

```python
df["Salary_vs_Department_Avg"] = (
    df["Salary"]
    - df["Department_Avg_Salary"]
)
```

```python
df["Salary_vs_Department_Avg"] = (
    df["Salary_vs_Department_Avg"]
    .round(2)
)
df
```

```python
df["Salary_Rank"] = (
    df["Salary"]
    .rank(
        ascending=False,
        method="dense"
    )
)
```

```python
df[
    ["Name", "Salary", "Salary_Rank"]
].sort_values("Salary_Rank")
```

```python
df["Skill_Rank"] = (
    df["Skill_Average"]
    .rank(
        ascending=False,
        method="dense"
    )
)
```

```python
df[
    ["Name", "Skill_Average", "Skill_Rank"]
].sort_values("Skill_Rank")
```

```python
df["Experience_Level"] = pd.cut(
    df["Experience"],
    bins=[0, 2, 5, 100],
    labels=[
        "Junior",
        "Mid-Level",
        "Senior"
    ]
)
```

```python
df[
    ["Name", "Experience", "Experience_Level"]
]
```

```python
df["Salary_Band"] = pd.cut(
    df["Salary"],
    bins=[0, 60000, 90000, 150000],
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)
```

```python
df[
        [
            "Name",
            "Salary",
            "Salary_Band"
        ]
    ]
```
