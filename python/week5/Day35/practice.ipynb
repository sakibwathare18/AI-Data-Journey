```python
import pandas as pd
```

```python
df = pd.read_csv("sales_data.csv")
```

```python
pivot = pd.pivot_table(
    df,
    index = "Region",
    columns = "Category",
    values = "Sales",
    aggfunc = "sum"
)
pivot
```

```python
pivot = pd.pivot_table(
    df,
    index = "Region",
    columns = "Category",
    values = "Sales",
    aggfunc = "mean"
)
pivot
```

```python
pivot = pd.pivot_table(
    df,
    index = "Region",
    columns = "Category",
    values = "Quantity",
    aggfunc = "sum"
)
pivot
```

```python
pivot = pd.pivot_table(
    df,
    index="Region",
    columns="Category",
    values=["Sales", "Rating"],
    aggfunc={"Sales": "sum", "Rating": "mean"},  # Maps specific functions to each column
    margins=True,
)
pivot
```

```python
region_order = pd.crosstab(
    df["Region"],
    df["Product"]
)
region_order
```

```python
region_order = pd.crosstab(
    df["Region"],
    df["Product"],
    normalize = "index"
)
region_order
```

```python
executive_matrix = pd.pivot_table(
    df,
    index=["Product", "Region"],
    values=["Sales", "Quantity", "Rating"],
    aggfunc={"Sales": "sum", "Quantity": "sum", "Rating": "mean"},
).reset_index()
executive_matrix
```

```python
executive_matrix.columns = [
    "Product",
    "Region",
    "Total Quantity",
    "Average Rating",
    "Total Sales",
]
```

```python
highest_selling = executive_matrix.loc[executive_matrix["Total Sales"].idxmax()]
lowest_selling = executive_matrix.loc[executive_matrix["Total Sales"].idxmin()]
```

```python
print("--- INSIGHTS ---")
print(
    f"1. Highest-selling combination:\nProduct: {highest_selling['Product']}\nRegion: {highest_selling['Region']}\nSales: ₹{highest_selling['Total Sales']:,}\n"
)
print(
    f"2. Lowest-selling combination:\nProduct: {lowest_selling['Product']}\nRegion: {lowest_selling['Region']}\nSales: ₹{lowest_selling['Total Sales']:,}\n"
)
```

```python
executive_matrix.to_csv("executive_sales_matrix.csv", index=False)
```

```python
region_summary = (
    pd.pivot_table(
        df,
        index="Region",
        values=["Sales", "Quantity", "Rating"],
        aggfunc={
            "Sales": ["sum", "mean"],
            "Quantity": "sum",
            "Rating": "mean",
        },
    )
    .reset_index()
)
```

```python
region_summary.columns = [
    "Region",
    "Total Quantity",
    "Average Rating",
    "Average Sales",
    "Total Sales",
]
```

```python
region_summary = region_summary[
    ["Region", "Total Sales", "Average Sales", "Total Quantity", "Average Rating"]
]
```

```python
region_summary.to_csv("region_summary.csv", index=False)
```
