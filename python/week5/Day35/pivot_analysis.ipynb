```python
import pandas as pd
```

 -------------------------
Load data


```python
df = pd.read_csv("sales_data.csv")
```

 -------------------------
 Region sales

```python
region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n===== REGION SALES =====")
region_sales
```

 -------------------------
 Pivot: Region × Category

```python
region_category = pd.pivot_table(
    df,
    index="Region",
    columns="Category",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)

print("\n===== REGION × CATEGORY =====")
region_category
```

 -------------------------
 Pivot with total

```python
region_category_total = pd.pivot_table(
    df,
    index="Region",
    columns="Category",
    values="Sales",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

print("\n===== REGION × CATEGORY WITH TOTALS =====")
region_category_total
```

 -------------------------
 Product × Region

```python
product_region = pd.pivot_table(
    df,
    index="Product",
    columns="Region",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)

print("\n===== PRODUCT × REGION =====")
product_region
```

 -------------------------
 Order count using crosstab

```python
order_count = pd.crosstab(
    df["Region"],
    df["Category"],
    margins=True
)

print("\n===== ORDER COUNT =====")
order_count
```

 -------------------------
 Percentage distribution

```python
percentage = pd.crosstab(
    df["Region"],
    df["Category"],
    normalize="index"
)

print("\n===== CATEGORY PERCENTAGE BY REGION =====")
percentage
```

 -------------------------
 Best combinations

```python
ranking = (
    region_category
    .stack()
    .sort_values(ascending=False)
)

print("\n===== BEST REGION-CATEGORY COMBINATIONS =====")
ranking
```

 -------------------------
 Export

```python
region_category_total.to_csv(
    "region_category_sales.csv"
)

print("\nReport exported successfully.")
```
