```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Customer_ID": [
        101, 102, 103, 104, 105, 105, 106, 107
    ],

    "Name": [
        " Amit ", "SNEHA", " Rahul",
        "Priya ", "vikas", "vikas",
        " NEHA ", "Arjun"
    ],

    "City": [
        "Pune", "mumbai", "PUNE",
        "Delhi", "Mumbai", "Mumbai",
        None, "Delhi"
    ],

    "Age": [
        24, 27, np.nan,
        31, 29, 29,
        26, np.nan
    ],

    "Salary": [
        50000, 65000, 55000,
        np.nan, 72000, 72000,
        60000, 48000
    ]
}

df = pd.DataFrame(data)
```

```python
df = df.drop_duplicates()
```

4. Clean Names

```python
df["Name"] = df["Name"].str.strip().str.title()
```

```python
df
```

Clean Cities

```python
df["City"] = df["City"].str.strip().str.title()
```

```python
df
```

Handle City

```python
df["City"] = df["City"].ffill()
```

```python
df
```

Handle Age

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df
```

Handle Salary

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
```

```python
df
```

Data Types

```python
df["Age"] = df["Age"].astype(int)
```

Bonus

Which city has the highest average salary?

```python
highest_avg_salary = df.groupby("City")["Salary"].mean().idxmax()
```

```python
highest_avg_salary
```

Which city has the most customers?

```python
city_count = df.groupby("City")["Customer_ID"].count()

most_customer_city = city_count.idxmax()

print(most_customer_city)
```

Who is the highest-paid customer?

```python
highest_paid_cust = df["Salary"].idxmax()

print(df.loc[highest_paid_cust, "Name"])
```

What is the average age?

```python
print(round(df["Age"].mean(), 2))
```

Final Report:

```python
print("=" * 50)
print("CUSTOMER DATA REPORT")
print("=" * 50)

total_cust = df["Customer_ID"].count()
avg_age = round(df["Age"].mean(), 2)
avg_salary = round(df["Salary"].mean(), 2)

print("\nTotal Customers:",total_cust)
print("Average Age:",avg_age)
print("Average Salary:",avg_salary)

highest_salary = df["Salary"].max()
highest_paid_cust = df["Salary"].idxmax()
cust_name = df.loc[highest_paid_cust, "Name"]

print("\nHighest Salary:",highest_salary)
print("Highest Paid Customer:",cust_name)

highest_paying_city = df.groupby("City")["Salary"].sum().idxmax()
common_city = df.groupby("City")["Salary"].median().idxmax()

print("\nBest Paying City:",highest_paying_city)
print("Most Common City:",common_city)

missing_values = df.isna().sum()
duplicate_rows = df.duplicated().sum()

print("\nMissing Values:\n",missing_values)
print("Duplicate Rows:",duplicate_rows)
```
