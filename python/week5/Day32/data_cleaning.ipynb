```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 105],
    "Name": [" Amit ", "Sneha", "RAHUL", "Priya ", "Vikas", "Vikas"],
    "Department": [
        "Data", "data", "IT", "HR", "IT", "IT"
    ],
    "Salary": [
        70000, 85000, np.nan, 55000, 75000, 75000
    ],
    "Experience": [
        2, 3, np.nan, 2, 3, 3
    ]
}

df = pd.DataFrame(data)
```

```python
df
```

Detect Missing Values

```python
df.isna()
```

Count Missing Values

```python
df.isna().sum()
```

isnull()

```python
df.isnull().sum()
```

Find Rows Containing Missing Values

```python
df[df.isna().any(axis=1)]
```

Find Rows With No Missing Values

```python
df[df.notna().all(axis=1)]
```

notna() → is this value present?\
all(axis=1) → are ALL values in this row present?

Remove Rows With Missing Data

```python
clean_df = df.dropna()
print(clean_df)
```

Remove Only Rows Where Specific Column Is Missing

```python
clean_df = df.dropna(
    subset=["Salary"]
)
clean_df
```

Fill Missing Values

```python
average_salary = df["Salary"].mean()
print(average_salary)
```

```python
df["Salary"] = df["Salary"].fillna(
    average_salary
)
```

```python
df
```

Fill With Median

```python
median_experience = df["Experience"].median()
df["Experience"] = df["Experience"].fillna(median_experience)
df
```

Forward Fill

```python
df["Salary"] = df["Salary"].ffill()
df
```

Backward Fill

```python
df["Salary"] = df["Salary"].bfill()
df
```

Find Duplicate Rows

```python
df.duplicated()
```

Count Duplicates

```python
df.duplicated().sum()
```

Remove Duplicates

```python
df = df.drop_duplicates()
df
```

Duplicate Based on One Column

```python
print(
    df.duplicated(
        subset=["Employee_ID"]
    )
)
```

Clean Text — str.strip()

```python
df["Name"] = df["Name"].str.strip()
df
```

Standardize Capitalization

```python
df["Name"] = df["Name"].str.title()
df
```

Standardize Department

```python
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
)
df
```

Replace Values

```python
df["Department"] = df["Department"].replace({
    "data": "Data",
    "DATA": "Data",
    "it": "IT"
})
```

Check Unique Values

```python
df["Department"].unique()
```

```python
df["Department"].nunique()
```

Data Types

```python
df.dtypes
```

Convert Data Type

```python
df["Experience"] = df["Experience"].astype(int)
```

```python
df.dtypes
```

Rename Columns

```python
df = df.rename(
    columns = {
        "Employee_ID" : "EmployeeID"
    }
)
df
```

Detect Invalid Values\
Suppose salary contains:\
70000\
85000\
-5000\

```python
invalid_salary = df[
    df["Salary"] < 0
]
print(invalid_salary)
```
