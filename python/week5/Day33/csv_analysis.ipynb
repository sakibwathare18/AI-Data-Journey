 -------------------------
 5. Calculated Column
 -------------------------

```python
df["Sales_Per_Unit"] = df["Sales"] / df["Quantity"]
df["Sales_Per_Unit"] = df["Sales_Per_Unit"].round(2)
```

```python
print(df[["Product", "Sales", "Quantity", "Sales_Per_Unit"]])
```

GroupBy — Real Analysis

Total sales by region:

```python
region_sales = df.groupby("Region")["Sales"].sum()
print(region_sales)
```

Total sales by category:

```python
category_sales = df.groupby("Category")["Sales"].sum()
print(category_sales)
```

Average rating by category:

```python
category_rating = df.groupby("Category")["Rating"].mean()
print(category_rating)
```

 -------------------------
 6. Region Analysis
 -------------------------

```python
report = df.groupby("Region").agg(
    Total_Sales = ("Sales", "sum"),
    Average_Sales = ("Sales", "mean"),
    Total_Quantity = ("Quantity", "sum"),
    Avearge_Rating = ("Rating", "mean")
)

print("\n===== REGION REPORT =====")
print(report)
```

 -------------------------
 7. Product Ranking
 -------------------------

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print(product_sales)
print(product_sales.head(1))
```
