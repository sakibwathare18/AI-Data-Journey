 -------------------------
 2. Filtering
 -------------------------

1. Sales greater than ₹50,000

```python
higher_sales = df[df["Sales"] > 50000]

print(higher_sales)
```

2. Rating greater than 4.5

```python
high_rating = df[df["Rating"] > 4.5]
print(high_rating)
```

3. Electronics products

```python
electronics = df[df["Category"] == "Electronics"]
print(electronics)
```

4. West region

```python
west = df[df["Region"] == "West"]
print(west)
```

5. Quantity greater than 3

```python
high_quantity = df[df["Quantity"] > 3]
print(high_quantity)
```

 -------------------------
 3. Calculations
 -------------------------

```python
df["Sales_Per_Unit"] = df["Sales"] / df["Quantity"]
df["Sales_Per_Unit"] = df["Sales_Per_Unit"].round(2)
```

```python
print("Highest Sales Per Unit:",df["Sales_Per_Unit"].max())
print("Lowest Sales Per Unit:",df["Sales_Per_Unit"].min())
print("Average Sales Per Unit:",round(df["Sales_Per_Unit"].mean(), 2))
```

 -------------------------
 4. GroupBy
 -------------------------

Total sales by Region

```python
df.groupby("Region")["Sales"].sum()
```

Total sales by Category

```python
df.groupby("Category")["Sales"].sum()
```

Average rating by Region

```python
df.groupby("Region")["Rating"].mean()
```

Total quantity by Product

```python
df.groupby("Product")["Quantity"].sum()
```

 -------------------------
 5. Ranking
 -------------------------

Top 3 products by total sales

```python
df.groupby("Product")["Sales"].sum().sort_values(ascending=False).head(3)
```

 -------------------------
 6. Regional Sales Report
 -------------------------

```python
report = df.groupby("Region").agg(
    Total_Sales = ("Sales", "sum"),
    Average_Sales = ("Sales", "mean"),
    Total_Quantity = ("Quantity", "sum"),
    Average_Rating = ("Rating", "mean"),
    Number_of_Orders = ("Order_ID", "count")
)
report
```

1. Sort by Total Sales

```python
report = report.sort_values("Total_Sales", ascending=False)
report
```

2. Export it

```python
report.to_csv("regional_sales_report.csv")
```
