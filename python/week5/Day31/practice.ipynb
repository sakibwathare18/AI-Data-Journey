```python
import pandas as pd
```

```python
data = {
    "Employee": [
        "Amit", "Sneha", "Rahul", "Priya", "Vikas",
        "Neha", "Arjun", "Pooja", "Karan", "Riya"
    ],

    "Department": [
        "Data", "Data", "IT", "HR", "IT",
        "Data", "HR", "IT", "Data", "IT"
    ],

    "City": [
        "Pune", "Mumbai", "Pune", "Delhi", "Mumbai",
        "Pune", "Delhi", "Mumbai", "Pune", "Delhi"
    ],

    "Salary": [
        70000, 85000, 65000, 55000, 75000,
        90000, 60000, 80000, 72000, 68000
    ],

    "Experience": [
        2, 3, 1, 2, 3,
        4, 2, 3, 2, 1
    ]
}

df = pd.DataFrame(data)
```

```python
df
```

1. Find the average salary by department.

```python
df.groupby("Department")["Salary"].mean()
```

2. Find the maximum salary by department.

```python
df.groupby("Department")["Salary"].max()
```

3. Find the minimum salary by department.

```python
df.groupby("Department")["Salary"].min()
```

4. Count the number of employees in each department.

```python
df.groupby("Department")["Employee"].count()
```

5. Calculate:\
Total Salary\
Average Salary\
Minimum Salary\
Maximum Salary

```python
df.groupby("Department").agg(
    Total_Salary=("Salary", "sum"),
    Average_Salary=("Salary", "mean"),
    Minimum_Salary=("Salary", "min"),
    Maximum_Salary=("Salary", "max"),
)
```

6. Find average salary by city.

```python
df.groupby("City")["Salary"].mean()
```

7. Find average salary by:\
Department + City

```python
df.groupby(["Department","City"])["Salary"].mean()
```

8. Find the department with the highest total salary.

```python
department_salary = df.groupby("Department")["Salary"].sum().idxmax()
print(department_salary)
```

9. Find the department with the highest average salary.

```python
department_avg_salary = df.groupby("Department")["Salary"].mean().idxmax()
print(department_avg_salary)
```

10. Create a department report:\
Department\
Total Employees\
Total Salary\
Average Salary\
Average Experience\
Maximum Salary

```python
department_report = (
    df.groupby("Department")
    .agg(
        Total_Employees = ("Salary", "count"),
        Total_Salary = ("Salary", "sum"),
        Average_Salary = ("Salary", "mean"),
        Average_Experience = ("Experience", "mean"),
        Maximum_Salary = ("Salary", "max")
    )
)

print(department_report)
```

Which department has the highest total salary, at least 3 employees, and an average salary above ₹70,000?

```python
dept_summary = df.groupby("Department").agg(
    Total_Employees = ("Salary", "count"),
    Total_Salary = ("Salary", "sum"),
    Average_Salary = ("Salary", "mean")
)
```

```python
filter = (dept_summary["Total_Employees"] > 3) & (dept_summary["Average_Salary"] > 70000)
```

```python
print(dept_summary[filter].sort_values("Total_Salary", ascending=False))
```
