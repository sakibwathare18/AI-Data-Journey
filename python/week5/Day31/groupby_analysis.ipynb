```python
import pandas as pd
```

```python
data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010,
                 1011, 1012],

    "Product": [
        "Laptop", "Phone", "Laptop", "Tablet",
        "Phone", "Monitor", "Laptop", "Tablet",
        "Phone", "Monitor", "Laptop", "Phone"
    ],

    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics"
    ],

    "Region": [
        "West", "South", "West", "North",
        "South", "West", "North", "North",
        "South", "West", "South", "North"
    ],

    "Sales": [
        65000, 30000, 72000, 25000,
        35000, 18000, 68000, 28000,
        32000, 20000, 70000, 34000
    ],

    "Quantity": [
        2, 1, 2, 1,
        2, 3, 2, 1,
        2, 2, 2, 1
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

groupby()\
How much total sales did each region generate?

```python
df.groupby("Region")["Sales"].sum()
```

GroupBy + Mean\
Average sales per region:

```python
regional_average = df.groupby("Region")["Sales"].mean()
print(regional_average)
```

GroupBy + Count\
How many orders came from each region?

```python
order_count = df.groupby("Region")["Order_ID"].count()
print(order_count)
```

Multiple Aggregations\
Suppose we want:\
total sales\
average sales\
minimum sales\
maximum sales

```python
region_analysis = df.groupby("Region")["Sales"].agg(
    ["sum", "mean", "min", "max"]
)
print(region_analysis)
```

Rename Aggregation Columns

```python
region_analysis = df.groupby("Region")["Sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean",
    Minimum_Sale="min",
    Maximum_Sale="max"
)
print(region_analysis)
```

Grouping Multiple Columns

```python
result = df.groupby(
    ["Region", "Product"]
)["Sales"].sum()
print(result)
```

reset_index()\
After grouping, you may get a Series or a DataFrame with grouped columns as the index.

```python
result = result.reset_index()
print(result)
```

GroupBy Quantity\
Total quantity sold per product:

```python
product_quantity = df.groupby(
    "Product"
)["Quantity"].sum()
print(product_quantity)
```

Total Sales by Product

```python
product_sales = df.groupby(
    "Product"
)["Sales"].sum()

print(product_sales)
```

```python
best_product = product_sales.idxmax()

print("Best Product:", best_product)
print("Sales:", product_sales.max())
```

Best and Worst Region

```python
regional_sales = df.groupby(
    "Region"
)["Sales"].sum()

best_region = regional_sales.idxmax()
worst_region = regional_sales.idxmin()

print("Best Region:",best_region)
print("Worst Region:",worst_region)
```

GroupBy + Multiple Columns

```python
region_report = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

print(region_report)
```

Sort Grouped Results

```python
region_report = region_report.sort_values(
    "Total_Sales",
    ascending=False
)

print(region_report)
```

Percentage Contribution

```python
region_report["Sales_Percentage"] = (
    region_report["Total_Sales"]
    / region_report["Total_Sales"].sum()
    * 100
)
```

```python
region_report["Sales_Percentage"] = (
    region_report["Sales_Percentage"].round(2)
)
print(region_report)
```

GroupBy With a Condition\
Suppose we only want orders where sales are greater than ₹30,000.

```python
high_sales = df[df["Sales"] > 30000]

result = high_sales.groupby(
    "Region"
)["Sales"].sum()

print(result)
```

GroupBy + Sorting = Business Analysis\
Let's answer:\
"Which products generated the most revenue?"

```python
product_report = df.groupby("Product").agg(
    Total_Sales=("Sales", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Average_Sale=("Sales", "mean")
)
```

```python
product_report = product_report.sort_values(
    "Total_Sales",
    ascending=False
)
```
