```python
import pandas as pd
```

```python
data = {
    "Product": [
        "Laptop", "Phone", "Tablet", "Monitor",
        "Keyboard", "Mouse", "Laptop", "Phone",
        "Monitor", "Tablet"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Accessories", "Accessories",
        "Electronics", "Electronics",
        "Electronics", "Electronics"
    ],
    "Price": [
        65000, 30000, 25000, 18000, 2500,
        1200, 72000, 35000, 20000, 28000
    ],
    "Quantity": [
        2, 5, 3, 4, 10,
        15, 1, 4, 2, 3
    ],
    "Rating": [
        4.5, 4.2, 4.0, 4.4, 4.1,
        3.9, 4.7, 4.3, 4.5, 4.1
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

Price\
Rating

```python
df[["Product","Price","Rating"]]
```

```python
df[df["Price"] > 25000]
```

4.Find products with:\

```python
df[df["Rating"] > 4.5]
```

5.Find products where:\
Price > 20000\
AND Rating >= 4.3

```python
df[
    (df["Price"] > 20000) &
    (df["Rating"] > 4.3)
]
```

6. Sort products by price from highest → lowest.

```python
df.sort_values(
    "Price",
    ascending=False
)
```

7. Create:\
Total_Sales = Price × Quantity

```python
df["Total_Sales"] = df["Price"] * df["Quantity"]
```

11. Find the number of products sold in each category.

```python
print(df.groupby("Category")["Quantity"].sum())
```

12. Sort the entire dataset by Total_Sales and display the top 3 products.

```python
df.sort_values("Total_Sales",ascending=False).head(3)
```

Which products generated more than ₹100,000 in total sales, have a rating of at least 4.3, and should be displayed from highest sales to lowest?

```python
result = df[
    (df["Total_Sales"] > 100000) &
    (df["Rating"] >= 4.3)
].sort_values("Total_Sales",ascending=False)
print(result)
```
