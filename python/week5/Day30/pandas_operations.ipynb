```python
import pandas as pd
```

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": [
        "Amit", "Sneha", "Rahul", "Priya",
        "Vikas", "Neha", "Arjun", "Pooja"
    ],
    "Department": [
        "Data", "Data", "IT", "HR",
        "IT", "Data", "HR", "IT"
    ],
    "Python": [85, 92, 76, 88, 69, 95, 73, 81],
    "SQL": [78, 89, 82, 91, 74, 96, 79, 85],
    "ML": [81, 95, 72, 86, 67, 94, 70, 88],
    "Experience": [1, 2, 1, 3, 2, 4, 1, 3]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

Selecting Columns

```python
print(df["Name"])
```

```python
print(df[["Name","Department","Python"]])
```

Selecting Rows

```python
print(df.iloc[0])
```

```python
print(df.iloc[0:3])
```

```python
print(df.iloc[[0, 2, 5]])
```

Filtering

```python
result = df[df["Python"] > 80]
print(result)
```

```python
result = df[
    (df["Python"] > 80) &
    (df["SQL"] > 80)
]
print(result)
```

```python
result = df[
    (df["Python"] > 90) |
    (df["ML"] > 90)
]

print(result)
```

```python
result = df[df["Department"] != "Data"]

print(result)
```

Filtering Text

```python
print(df[df["Department"] == "IT"])
```

```python
print(df[df["Department"] == "Data"])
```

isin()

```python
result = df[
    df["Department"].isin(["Data","IT"])
]
print(result)
```

between()

```python
result = df[
    df["Experience"].between(2,3)
]
print(result)
```

Sorting Data

```python
sorted_df = df.sort_values("Python")
print(sorted_df)
```

```python
sorted_df = df.sort_values(
    "Python",
    ascending=False
)
print(sorted_df)
```

Sort by Multiple Columns

```python
sorted_df = df.sort_values(
    ["Department", "Python"],
    ascending=[True, False]
)

print(sorted_df)
```

Adding a New Column

```python
df["Average"] = (
    df["Python"] +
    df["SQL"] +
    df["ML"]
) / 3
print(df)
```

```python
df["Average"] = df["Average"].round(2)
print(df)
```

Create a Performance Category

```python
def performance_category(avg):

    if avg >= 85:
        return "Excellent"
    elif avg >= 75:
        return "Good"
    elif avg >= 60:
        return "Average"
    else:
        return "Needs Improvement"
```

```python
df["Performance"] = df["Average"].apply(performance_category)
print(df)
```

Show me Data department employees with an average above 80, sorted from highest to lowest.

```python
result = df[
    (df["Department"] == "Data") &
    (df["Average"] > 80)
].sort_values(
    "Average",
    ascending=False
)
print(result)
```
