```python
import pandas as pd
```

--------
Task 1

```python
customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")
products = pd.read_csv("products.csv")
```

```python
print("Customers:")
print("\nShape:",customers.shape)
print("Columns:",customers.columns)
print("5 Rows:",customers.head())
```

```python
print("Orders:")
print("\nShape:",orders.shape)
print("Columns:",orders.columns)
print("5 Rows:",orders.head())
```

```python
print("Products:")
print("\nShape:",products.shape)
print("Columns:",products.columns)
print("5 Rows:",products.head())
```

--------
Task 2

```python
pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="inner"
)
```

```python
customer_orders = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="left"
)
customer_orders
```

```python
pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="right"
)
```

```python
pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="outer"
)
```

------------
Task 3

```python
product_order = pd.merge(
    orders,
    products,
    on="Product_ID",
    how="left"
)
product_order
```

-------
Task 4

```python
product_order["Total_Sales"] = product_order["Quantity"] * product_order["Price"]
```

---------
Task 5

Top customer by spending

```python
product_order.groupby("Customer_ID")["Total_Sales"].sum().sort_values(ascending=False)
```

Top product by sales

```python
product_order.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
```

Top city by sales

```python
full_data = pd.merge(
    customers,
    product_order,
    on="Customer_ID",
    how="left"
)
```

```python
full_data.groupby("City")["Total_Sales"].sum().sort_values(ascending=False)
```

---------
Task 6

```python
no_order = full_data[full_data["Order_ID"].isna()]
no_order
```

--------
Task 7\
Boss Challenge\
Final Customer Sales Report

```python
report = full_data.groupby(
    ["Customer_ID","Name","City"]
).agg(
    Total_Orders = ("Order_ID", "count"),
    Total_Quantity = ("Quantity", "sum"),
    Total_Sales = ("Total_Sales", "sum")
)
```

```python
report["Average_Order_Value"] = (
    report["Total_Sales"]
    / report["Total_Orders"]
)
```

```python
report = report.sort_values(
    "Total_Sales",
    ascending=False
)
```
