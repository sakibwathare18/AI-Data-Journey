```python
import pandas as pd
```

 -------------------------
Load Datasets

```python
customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")
products = pd.read_csv("products.csv")

print("CUSTOMERS")
print(customers)
```

```python
print("\nORDERS")
print(orders)
```

```python
print("\nPRODUCTS")
print(products)
```

Merge

```python
customer_order = pd.merge(
    customers,
    orders,
    on="Customer_ID"
)
print(customer_order)
```

Inner Join

```python
pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="inner"
)
```

 -------------------------
 Merge customers + orders


```python
customer_orders = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="left"
)
print(customer_orders)
```

Right Join

```python
result = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="right"
)
result
```

Right Join

```python
result = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="outer"
)
result
```

 -------------------------
 Merge orders + products

```python
order_products = pd.merge(
    orders,
    products,
    on="Product_ID",
    how="left"
)
order_products
```

 -------------------------
 Calculate sales

```python
order_products["Total_Sales"] = (
    order_products["Quantity"]
    * order_products["Price"]
)
```

```python
print(
    order_products[
        [
            "Order_ID",
            "Product",
            "Quantity",
            "Price",
            "Total_Sales"
        ]
    ]
)
```

 -------------------------
 Full dataset

```python
customer_orders = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="left"
)
```

```python
full_data = pd.merge(
    customer_orders,
    products,
    on="Product_ID",
    how="left"
)
```

Calculate revenue:

```python
full_data["Total_Sales"] = (
    full_data["Quantity"]
    * full_data["Price"]
)
full_data
```

 -------------------------
 Customer analysis

```python
customer_sales = (
    full_data
    .groupby(["Customer_ID", "Name"])["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)
customer_sales
```

 -------------------------
 City analysis

```python
city_sales = (
    full_data
    .groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)
city_sales
```

 -------------------------
 Product analysis

```python
product_sales = (
    full_data
    .groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)
product_sales
```

 -------------------------
 Find customers with no orders

Customers With No Orders

```python
customers_with_orders = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="left"
)
no_orders = customers_with_orders[
    customers_with_orders["Order_ID"].isna()
]
no_orders
```

 -------------------------
 Export

```python
full_data.to_csv(
    "full_sales_data.csv",
    index=False
)
```
